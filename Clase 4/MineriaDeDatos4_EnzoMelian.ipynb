```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Se crea el dataframe
df = pd.DataFrame({
"Cliente":
 [1, 2, 3, 4, 5, 6, 7, 8],
"Edad":
 [25, 40, 35, 50, 30, 45, 38, 29],
"Ingresos":
 [50000, 60000, 70000, 80000, 45000, 85000, 72000, 48000],
"Compras":
 [5, 15, 10, 20, 7, 18, 12, 6],
"Categoría": ["A", "B", "C", "A", "B", "C", "A", "B"]
})

df
```

## 1. Análisis Exploratorio de Datos (EDA)


 #### Histograma de edad

```python
sns.histplot(df["Edad"], bins=5, kde=True)
plt.title("Distribución de Edades")
plt.show()
```

 #### Histograma de ingreso

```python
sns.histplot(df["Ingresos"], bins=5, kde=True)
plt.title("Distribución de ingresos")
plt.show()
```

 #### Boxplot para visualizar valores atipicos en compras

```python
sns.boxplot(x=df["Compras"])
plt.title("Boxplot de Compras")
plt.show()
```

Generar scatter plot para analizar la relacion entre ingresos y compras

```python
sns.scatterplot(x=df["Ingresos"], y=df["Compras"])
plt.title("Relación entre Ingresos y Compras")
plt.show()
```

## 2. Visualización Avanzada

Crear un Pairplot con las variables numéricas.

```python
sns.pairplot(df[["Edad", "Ingresos", "Compras"]])
plt.show()
```

Generar un Heatmap de correlaciones entre todas las variables numéricas.

```python
correlaciones = df.corr(numeric_only=True)
sns.heatmap(correlaciones, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Mapa de Correlación de Variables")
plt.show()
```

Utilizar un Violin Plot para comparar la distribución de Compras según la Categoría.

```python
sns.violinplot(x="Categoría", y="Compras", data=df, hue="Categoría", legend=False)
plt.title("Distribución de Compras por Categoría")
plt.show()
```
